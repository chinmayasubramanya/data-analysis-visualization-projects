```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\chinm\OneDrive\Desktop\data-analysis-visualization-projects\02-netflix_library_explorer\netflix_titles.csv",
    encoding="latin1"
)

print("Dataset loaded successfully!")

print("\nSample Data (first 5 rows):")
print(df[["title", "type", "release_year"]].head())

print("\n--- Dataset Overview ---")
print("Shape of the dataset:", df.shape)

print("Column Names:")
print(list(df.columns))

print("\n--- Content Type Distribution ---")
content_counts = df["type"].value_counts()
print(content_counts)

recent_content = df[df["release_year"] > 2015]

print("\n--- Content Released After 2015 ---")
print(recent_content[["title", "type", "release_year"]].head())

print("\n--- Top 5 Content Ratings ---")
ratings = df["rating"].value_counts().head(5)
print(ratings)

sorted_df = df.sort_values(
    by="release_year",
    ascending=False
)

print("\n--- Latest Content on Netflix ---")
print(sorted_df[["title", "type", "release_year"]].head())
```
